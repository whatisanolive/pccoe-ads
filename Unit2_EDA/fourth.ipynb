```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {'Transaction_ID': [1, 2, 3, 4, 5],
        'Quantity': [2, 1, 3, 4, 2],
        'Price_per_item': [10.50, 25.00, 5.75, 12.00, 15.50],
        'Customer_ID': ['C001', 'C002', 'C001', 'C003', 'C002']}
```

```python
df = pd.DataFrame(data)
df
```

```python
df['total_price'] = df['Quantity']*df['Price_per_item']
df
```

```python
df['Transaction_date'] = pd.to_datetime(['2024-01-23','2024-02-13','2024-07-19','2024-04-29','2024-08-12'])
df
```

```python
df['Transaction_day'] = df['Transaction_date'].dt.day_of_week
df['Transaction_month'] = df['Transaction_date'].dt.month
df['Transaction_year'] = df['Transaction_date'].dt.year
df
```

```python
df = df.drop(columns=['Transaction_data'])
```

```python
df
```

```python
df_encoded = pd.get_dummies(df, columns=['Customer_ID'], prefix='Customer', dtype=np.int32)

print( "\n df with one hot endoded customer id")
df_encoded
```
