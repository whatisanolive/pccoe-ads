```python
sns.boxplot(x=df['spending_amount'])
```

```python
sns.histplot(df['spending_amount'], bins=20,kde=True)
plt.title("Customer spending")
```

```python
lower_treshold = df['spending_amount'].quantile(0.05)
upper_treshold = df['spending_amount'].quantile(0.95)
```

```python
print(lower_treshold,upper_treshold)
```

```python
outliers = df[(df['spending_amount'] < lower_treshold) | (df['spending_amount'] > upper_treshold)]
outliers.head()
```

```python
df_cleaned = df[(df['spending_amount'] >= lower_treshold) & (df['spending_amount'] <= upper_treshold)]
df_cleaned.info()
```

```python
sns.histplot(df_cleaned['spending_amount'], bins=20,kde=True)
plt.title("Customer spending")
```
