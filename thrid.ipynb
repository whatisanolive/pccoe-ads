```python
x = ['sun','sat','mon','fri','tue','wed','thu']

y = [5,6.7,4,6,2,4.9,1.8]
df3=pd.DataFrame({'days':x, 'amount':y})

ax = sns.stripplot(df3)

ax.set(xlabel='days',ylabel='amount')

plt.title('expenses')
plt.show()
```
