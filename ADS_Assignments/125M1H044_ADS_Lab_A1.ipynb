```python
# Question 1(b-v) — Combined: Create / Access / Update data structures using loops and control structures
# Write your code below

users_list = ["Alice", "Bob", "Charlie"]

roles_tuple = ("Admin", "Editor", "Viewer")


user_profiles = {
    "Alice": {"active": True, "logins": 5},
    "Bob": {"active": False, "logins": 12},
    "Charlie": {"active": True, "logins": 3}
}

banned_users = {"Dave", "Eve"}

i = 0
# Outer Loop: Iterates through the users list
while i < len(users_list):
    current_user = users_list[i]
    print(f"\nProcessing User: {current_user}")
    
    j = 0
    # Inner Loop: Iterates through the available roles tuple
    while j < len(roles_tuple):
        current_role = roles_tuple[j]
        
        # Control Structure: Accessing & checking Dictionary data
        is_active = user_profiles[current_user]["active"]
        login_count = user_profiles[current_user]["logins"]
        
        if is_active:
            # Control Structure: Check if user qualifies for a role level based on logins
            if current_role == "Admin" and login_count >= 10:
                print(f"  Assigned Role: {current_role}")
            elif current_role == "Editor" and login_count >= 4:
                print(f"  Assigned Role: {current_role}")
                # Updating Dictionary: Modify login milestones dynamically
                user_profiles[current_user]["logins"] += 1 
            elif current_role == "Viewer":
                print(f"  Assigned Role: {current_role} (Default)")
                
        else:
            # Control Structure: Flag inactive accounts and add them to the Banned Set
            print(f" -> Account inactive! Adding {current_user} to restriction watch.")
            banned_users.add(current_user)  # Updating Set
            break  # Break inner loop early if account is inactive
            
        j += 1 
        
    i += 1  


print("\nFinal Data Structure States")
print(f"List (Unchanged): {users_list}")
print(f"Tuple (Unchanged): {roles_tuple}")
print(f"Updated Dictionary: {user_profiles}")
print(f"Updated Set (Bob added): {banned_users}")

# Expected Output: Combined demonstration using all four data structures
```

### Question 2 - Central Tendency: Mean, Median and Mode

Write a Python **function** to compute the **mean**, **median**, and **mode** of a list of numbers.

**Requirements:**
- Define separate functions: `compute_mean(data)`, `compute_median(data)`, `compute_mode(data)`.
- Do **not** use the `statistics` module — implement the logic manually.
- Test your functions with at least two different datasets (one with a unique mode, one with multiple modes).
- Print the result for each measure clearly.

> _Formula reference:_
> - Mean: $\bar{x} = \dfrac{\sum x_i}{n}$
> - Median: Middle value of the sorted list (average of two middle values if $n$ is even)
> - Mode: Value(s) that appear most frequently

```python
# Question 2 — Mean, Median and Mode
# Write your code below

def compute_mean(data):
    if not data:
        return 0
    return sum(data) / len(data)


def compute_median(data):
    if not data:
        return None

    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2


    if n%2 == 0:
        return (sorted_data[mid-1] + sorted_data[mid]) / 2
    
    else:

        return sorted_data[mid]



def compute_mode(data):

    if not data:
        return []
    

    frequency = {}
    for item in data:
        frequency[item] = frequency.get(item, 0) + 1
        

    max_count = max(frequency.values())
    

    modes = [key for key, count in frequency.items() if count == max_count]
    
    return modes


dataset_1 = [4, 1, 2, 2, 3, 5, 2]


dataset_2 = [10, 20, 20, 30, 40, 40, 50, 60]

print("Dataset 1 Test")
print(f"Data: {dataset_1}")
print(f"Sorted: {sorted(dataset_1)}")
print(f"Mean:   {compute_mean(dataset_1):.2f}")
print(f"Median: {compute_median(dataset_1)}")
print(f"Mode:   {compute_mode(dataset_1)}\n")

print("Dataset 2 Test")
print(f"Data: {dataset_2}")
print(f"Sorted: {sorted(dataset_2)}")
print(f"Mean:   {compute_mean(dataset_2):.2f}")
print(f"Median: {compute_median(dataset_2)}")
print(f"Mode:   {compute_mode(dataset_2)}")

# Expected Output:
# Mean   : <value>
# Median : <value>
# Mode   : <value(s)>
```

### Question 3 - Measures of Dispersion: Variance and Standard Deviation

Write a Python **function** to compute the **variance** and **standard deviation** of a list of numbers.

**Requirements:**
- Define functions: `compute_variance(data)` and `compute_std_dev(data)`.
- Implement both **population** and **sample** variants.
- Do **not** use the `statistics` or `numpy` module — implement the logic manually.
- Test your functions with at least two different datasets.

> _Formula reference:_
> - Population Variance: $\sigma^2 = \dfrac{\sum (x_i - \bar{x})^2}{n}$
> - Sample Variance: $s^2 = \dfrac{\sum (x_i - \bar{x})^2}{n - 1}$
> - Standard Deviation: $\sigma = \sqrt{\sigma^2}$

```python
# Question 3 — Variance and Standard Deviation
# Write your code below
def compute_mean(data):
    if not data:
        return 0
    return sum(data) / len(data)


def compute_variance(data, is_sample=False):
    
    if len(data) < 2 and is_sample:
        raise ValueError("Sample variance requires at least two data points.")
    if not data:
        return 0
        
    mean = compute_mean(data)
    
    # Calculate sum of squared differences from the mean
    squared_diff_sum = sum((x - mean) ** 2 for x in data)
    
    # Choose denominator based on population or sample configuration
    denominator = (len(data) - 1) if is_sample else len(data)
    
    return squared_diff_sum / denominator


def compute_std_dev(data, is_sample=False):
    
    variance = compute_variance(data, is_sample)
    return variance ** 0.5

dataset_1 = [10, 12, 23, 23, 16, 23, 21, 16]

# Dataset 2: Highly dispersed test set
dataset_2 = [4, 8, 15, 16, 23, 42]

print("Dataset 1 Test")
print(f"Data: {dataset_1}")
print(f"Population Variance: {compute_variance(dataset_1, is_sample=False):.4f}")
print(f"Sample Variance:     {compute_variance(dataset_1, is_sample=True):.4f}")
print(f"Population Std Dev:  {compute_std_dev(dataset_1, is_sample=False):.4f}")
print(f"Sample Std Dev:      {compute_std_dev(dataset_1, is_sample=True):.4f}\n")

print("Dataset 2 Test")
print(f"Data: {dataset_2}")
print(f"Population Variance: {compute_variance(dataset_2, is_sample=False):.4f}")
print(f"Sample Variance:     {compute_variance(dataset_2, is_sample=True):.4f}")
print(f"Population Std Dev:  {compute_std_dev(dataset_2, is_sample=False):.4f}")
print(f"Sample Std Dev:      {compute_std_dev(dataset_2, is_sample=True):.4f}")

# Expected Output:
# Population Variance : <value>
# Sample Variance     : <value>
# Population Std Dev  : <value>
# Sample Std Dev      : <value>
```

### Question 4 - Frequency Distribution Table

Consider an **ungrouped dataset of your choice** (minimum **100 data items**) and use statistical concepts to prepare a **grouped frequency distribution table** using Python.

**Step-by-step Tasks:**

| Sub-part | Task |
|---|---|
| **(a)** | Determine the **range** of the raw data: $\text{Range} = x_{\max} - x_{\min}$ |
| **(b)** | Determine the **number of classes** using Sturges formula: $k = 1 + 3.322 \log_{10} n$ |
| **(c)** | Determine the **class width**: $h = \lceil \text{Range} / k \rceil$ |
| **(d)** | Determine the **midpoint** of each class interval: $m = (\text{lower} + \text{upper}) / 2$ |
| **(e)** | Determine the **relative frequency** of each class: $f_{rel} = f / n$ |
| **(f)** | Determine the **cumulative frequency** of each class |
| **(g)** | Populate the complete **frequency distribution table** with columns: Class Interval, Frequency, Midpoint, Relative Frequency, Cumulative Frequency |

> _You may generate synthetic data using `random` or `numpy.random`, or use a real-world dataset._  
> _Display the final table using a Pandas DataFrame._

```python
# Question 4(a) — Determine the Range of the raw data
# Write your code below
import math
import numpy as np
import pandas as pd

# 1. Generate an ungrouped dataset of 100 items (using fixed seed for stable output)
np.random.seed(42)
raw_data = np.random.randint(31, 101, size=100).tolist()

n = len(raw_data)
x_min = min(raw_data)
x_max = max(raw_data)

print(f"Dataset of {n} items successfully generated.")
print("Data generated: ", raw_data)

range_val = x_max - x_min

print(f"Range = {x_max} - {x_min} = {range_val}")



# Expected Output: Range = <value>
```

```python
# Question 4(b) — Number of Classes (Sturges formula)
# Write your code below

k = int(math.ceil(1 + 3.322 * math.log10(n)))

print(f"Calculated number of classes (rounded up): {k}")

# Expected Output: Number of Classes (k) = <value>
```

```python
# Question 4(c) — Class Width
# Write your code below


h = math.ceil(range_val / k)

print(f"Class Width (h): {h}")


# Expected Output: Class Width (h) = <value>
```

```python
# Initialize interval boundaries
intervals = []
current_lower = x_min

for _ in range(k):
    current_upper = current_lower + h - 1
    intervals.append((current_lower, current_upper))
    current_lower += h

# Count how many data items fall into each class boundary
frequencies = [0] * k
for val in raw_data:
    for idx, (lower, upper) in enumerate(intervals):
        if lower <= val <= upper:
            frequencies[idx] += 1
            break

print("Interval buckets and base frequencies established successfully!")
print(f"Intervals: {intervals}")
print(f"Frequencies: {frequencies}")
```

```python
# Question 4(d) — Midpoint of each class interval
# Write your code below

midpoints_col = []

for lower, upper in intervals:
    midpoint = (lower + upper) / 2
    midpoints_col.append(midpoint)

print("Sub-part (d) Results - Midpoints calculated:")
for i, m in enumerate(midpoints_col):
    print(f"Class {i+1} Midpoint: {m}")

# Expected Output: List of class midpoints
```

```python
# Question 4(e) — Relative Frequency
# Write your code below

# (e) Determine the relative frequency of each class
relative_freq_col = []

for freq in frequencies:
    rel_freq = round(freq / n, 4)
    relative_freq_col.append(rel_freq)

print("Sub-part (e) Results - Relative Frequencies calculated:")
for i, rf in enumerate(relative_freq_col):
    print(f"Class {i+1} Relative Frequency: {rf}")


# Expected Output: List of relative frequencies (should sum to 1.0)
```

```python
# Question 4(f) — Cumulative Frequency
# Write your code below

cumulative_freq_col = []
running_cumulative_freq = 0

for freq in frequencies:
    running_cumulative_freq += freq
    cumulative_freq_col.append(running_cumulative_freq)

print("Sub-part (f) Results - Cumulative Frequencies calculated:")
for i, cf in enumerate(cumulative_freq_col):
    print(f"Class {i+1} Cumulative Frequency: {cf}")



# Expected Output: List of cumulative frequencies (last value should equal n)
```

```python
# Question 4(g) — Complete Frequency Distribution Table (Pandas DataFrame)
# Combine all sub-parts (a) to (f) into one complete table
# Write your code below
class_intervals_col = [f"{lower} - {upper}" for lower, upper in intervals]

# (g) Populate the complete frequency distribution table
df = pd.DataFrame({
    "Class Interval": class_intervals_col,
    "Frequency": frequencies,
    "Midpoint": midpoints_col,
    "Relative Frequency": relative_freq_col,
    "Cumulative Frequency": cumulative_freq_col
})

print("Sub-part (g) - Final Frequency Distribution Table:")
df
# Expected Output: DataFrame with columns:
#   Class Interval | Frequency | Midpoint | Relative Frequency | Cumulative Frequency
```
