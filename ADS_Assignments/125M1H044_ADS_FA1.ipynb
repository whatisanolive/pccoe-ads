### Univariate Analysis



## a) Explain univariate stats

Univariate analysis is the statistical analysis of a single variable. Its main purpose is to understand the characteristics and distribution of that variable.

It focuses on measures such as:

- Mean - average of all observations
- Median - middle most value
- Mode - most frequent value, it can be single modal or multi modal
- Variance - shows how far a set of data points are spread from mean
- Standard Deviation - average distance of data points from their mean
- Skewness - is the asymmetry of a data distribution around its mean
- Kurtosis - is the measure of thickness of tail in a data distribution
- Minimum and Maximum values - shows the range of data



```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# loading tips dataset
df = sns.load_dataset('tips')
df.head()
```

```python
print("Dataset shape")
df.shape
```

```python
print("Dataset info:")
df.info()
```

```python
print("Basic stats about the dataset:")
df.describe()
```

```python
# we'll be using total_bill as our variable - its a continous values which are in dollars

total_bill = df['total_bill']
total_bill.head()
```

## b) Compute mean, median, mode, variance, skewness, kurtosis.

```python
mean_bill = total_bill.mean()

print("Mean Total Bill:", round(mean_bill, 2))
```

```python
median_bill = total_bill.median()

print("Median Total Bill:", median_bill)
```

```python
mode_bill = total_bill.mode()

print("Mode of Total Bill:")
print(mode_bill.values)
```

```python
variance_bill = total_bill.var()

print("Variance of Total Bill:", round(variance_bill, 2))
```

```python
std_bill = total_bill.std()

print("Standard Deviation:", round(std_bill, 2))
```

```python
skewness_bill = total_bill.skew()

print("Skewness:", round(skewness_bill, 3))
```

```python
kurtosis_bill = total_bill.kurt()

print("Kurtosis:", round(kurtosis_bill, 3))
```

```python
summary = pd.DataFrame({
    'Statistic': [
        'Mean',
        'Median',
        'Mode',
        'Variance',
        'Standard Deviation',
        'Skewness',
        'Kurtosis',
        'Minimum',
        'Maximum'
    ],
    
    'Value': [
        total_bill.mean(),
        total_bill.median(),
        total_bill.mode().iloc[0],
        total_bill.var(),
        total_bill.std(),
        total_bill.skew(),
        total_bill.kurt(),
        total_bill.min(),
        total_bill.max()
    ]
})

summary
```

## c) Visualize with histogram + boxplot

#### Histogram - helps us to visualize the distribution of our data, so we can how much data falls in which bin (range), skewness of the data, outliers too.

```python
plt.figure(figsize=(9,5))

sns.histplot(
    data=df,
    x='total_bill',
    bins=10,
    kde=True
)

plt.title("Distribution of Total Restaurant Bills")
plt.xlabel("Total Bill ($)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.histplot(
    data=df,
    x='total_bill',
    bins=20,
    kde=True
)

plt.axvline(
    mean_bill,
    linestyle='--',
    linewidth=2,
    label=f'Mean = {mean_bill:.2f}'
)

plt.axvline(
    median_bill,
    linestyle=':',
    linewidth=2,
    label=f'Median = {median_bill:.2f}'
)

plt.title("Distribution of Total Bills with Mean and Median")
plt.xlabel("Total Bill ($)")
plt.ylabel("Frequency")

plt.legend()

plt.show()
```

#### Boxplot - gives us the midian, min ,max values and gives us outliers.

```python
plt.figure(figsize=(9, 4))

sns.boxplot(
    data=df,
    x='total_bill'
)

plt.title("Box Plot of Total Restaurant Bills")
plt.xlabel("Total Bill ($)")

plt.show()
```
